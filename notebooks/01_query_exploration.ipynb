```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
api_key = os.getenv("LENS_API_KEY")
url = "https://api.lens.org/patent/search"
headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json"
}
```

```python
body = {
    "query": {
        "bool": {
            "must": [
                {
                    "bool": {
                        "should": [
                            {"match_phrase": {"title": "bio-based adhesive"}},
                            {"match_phrase": {"title": "bio-adhesive"}},
                            {"match_phrase": {"title": "biobased adhesive"}},
                            {"match_phrase": {"title": "plant-based adhesive"}},
                            {"match_phrase": {"title": "natural adhesive"}}
                        ],
                        "minimum_should_match": 1
                    }
                },
                {"prefix": {"class_cpc.symbol": "C09J"}},
                {
                    "range": {
                        "date_published": {
                            "gte": "2011-01-01",
                            "lte": "2026-12-31"
                        }
                    }
                }
            ]
        }
    },
    "size": 5,
    "include": ["lens_id", "biblio.invention_title", "abstract", "date_published", "biblio.classifications_cpc.classifications.symbol"]
}
```

```python
response = requests.post(url, headers=headers, json=body)
print("Status code:", response.status_code)
response.json()
```
